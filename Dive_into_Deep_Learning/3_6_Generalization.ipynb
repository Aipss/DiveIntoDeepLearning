##### exercise_4
* K-fold cross-validation requires K times of training and each epoch needs different train set,
  which requires abundant space and compotaton to achieve.
* As this methord is usually used for select better model, it needs even more times of training and validation.

##### exercise_5
* The split has limits.
  + The distribution: Each training set and validation set may not follow the same distribution as the whole set.
  + K-fold cross-validaton methord will destruct the relation between samples.
* The overfitting and underfitting.
  + When each training set has smaller number of samples, the model may be overfitting.
  + When K is smaller, the validation may not enough and so that the model is underfitting.

##### exercise_6
