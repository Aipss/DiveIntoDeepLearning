#### exercise_1
According to Hoeffding:   
$P({\epsilon}_D(f) - \epsilon (f) \geq t) < e^{-2nt^2} $    
Thus, $P({\epsilon}_D(f) - \epsilon (f) < t) \geq e^{-2nt^2} $      
So, $t = 0.00001 $ and $e^{-2nt^2} = 0.999 $, the calculation result is $n = 5\times 10^6 $

The dimension is 6:   
It could be illustrated that a $x$-order polynomial has $x+1$ VC dimension.    
As a $x$-order polynomial has $x+1$ independent peremeters, its could fit $x+1$ different points.   
Consider $x+2$ different points, as the $x$-order polynomial has no more peremeters, it could not fit all of the points perfectly.    
In conclusion, a $x$-order polynomial has $x+1$ VC dimension.
