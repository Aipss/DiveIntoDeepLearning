##### exercise_10
It varies.  
Generally, producting a m*n and a n*k tensor requires n*(m*k) times of product and m*k times of summation.  
If (AB)C requires m*k*(n+t) times of production, A(BC) requires n*t*(m+k) times of production.  
So (AB)C (2^31) calculates quicker tnan A(BC) (2^40).  

##### exercise_11
The difference occurs when C is transposed. Commonly, A(C.T) is better.  
Commonly, a tensor is more likely to be memoried line by line.  
When calculating AB, B is accessed column by column, and it takes time.  
On the contrary, when calculating A(C.T), C is accessed line by line, which is constant, and it saves time.  
Without cloning memory, C and B will refer to same piece of place.  
It saves sapce, but C's change will influence B, vise versa.
