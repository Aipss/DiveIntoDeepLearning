#### 53. The bottom half of the parabola $x + (y-1)^2 = 0$

> Rewrite the equation $x + (y-1)^2 = 0$ in terms of $y$:
> $$ \begin{align*}
> x + (y-1)^2 &= 0 \\
> (y-1)^2 &= -x \\
> y-1 &= \pm \sqrt{-x} \\
> y &= 1 \pm \sqrt{-x}
> \end{align*} $$
> The bottom half is $y = 1 - \sqrt{-x}$, whose domain is $(-\infty, 0]$ and range is $(-\infty, 1]$. The graph of the function is shown below.

```python
x = np.linspace(-10, 0, 100)
y = 1 - np.sqrt(-x)

fig, ax = plt.subplots(figsize=(4, 3))
ax.plot(x, y)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True)
plt.show()
```

#### 79. If $f$ and $g$ are both even functions, if $f+g$ even? If $f$ and $g$ are both odd functions, if $f+g$ odd? What if $f$ is even and $g$ is odd? Justify your answers.

> - **If $f$ and $g$ are both even functions:** 
> 
>   We have $f(x) = f(-x)$ and $g(x) = g(-x)$.   
> 
>   Let $h = f + g$, then $h(-x) = f(-x) + g(-x) = f(x) + g(x) = h(x)$, which indicates that $h$, that is $f+g$, is even.

> - **If $f$ and $g$ are both odd functions**
> 
>   We have $f(x) = -f(-x)$ and $g(x) = -g(-x)$.  
>  
>   Let $h = f+g$, then $h(-x) = f(-x) + g(-x) = -f(x) - g(x) = -h(x)$, which indicates that $h$, that is $f+g$, is odd.

> - **If $f$ is even and $g$ is odd:**
> 
>   We have $f(x) = f(-x)$ and $g(x) = -g(-x)$.   
> 
>   Let $h = f+g$, then $h(-x) = f(-x) + g(-x) = f(x) - g(x)$.   
> 
>   Because $h(x) + h(-x) = 2f(x)$ and $h(x) - h(-x) = 2g(x)$, so $h$, that is $f+g$, is even AND odd if and only if $f(x) \equiv g(x) \equiv 0$.    
>
>   Notice that $h \equiv 0$ when $f \equiv g \equiv 0$.

#### 80. If $f$ and $g$ are both even functions, is the product $fg$ even? If $f$ and $g$ are both odd functions, is $fg$ odd? What if $f$ is even and $g$ is odd? Justify your answers.

> - **If $f$ and $g$ are both even functions:**
>
>   We have $f(x) = f(-x)$ and $g(x) = g(-x)$.
> 
>   Let $h = fg$, then $h(-x) = f(-x)g(-x) = f(x)g(x) = h(x)$.
>
>   So $h$, that is $fg$, is even.

> - **If $f$ and $g$ are both odd functions:**
> 
>   We have $f(x) = -f(-x)$ and $g(x) = -g(-x)$.
> 
>   Let $h = fg$, then $h(-x) = f(-x)g(-x) = [-f(x)][-g(x)]=f(x)g(x) = h(x)$.
> 
>   So $h$, that is $fg$, is even.

| Income | Ulcer rate (Per 100 population) |
| :--: | :--: |

##### (a) Make a scatter plot of these data and decide whether a linear model is appropriate.

```python
income = np.array([4, 6, 8, 12, 16, 20, 30, 45, 60]) * 1_000
rate = np.array([14.1, 13.0, 13.4, 12.5, 12.0, 12.4, 10.5, 9.4, 8.2])

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.scatter(x=income, y=rate)
ax.set_xlabel('Income ($)')
ax.set_ylabel('Ulcer Rate (Per 100 Population)')
ax.grid(True)
plt.show()
```

> According to the scatter plot above, we could see a line approximately.
> So we could give linear model a try.

##### (b) Find and graph a linear model using the first and last data points.

```python
# the line will go through the initial point (4_000, 14.1) and the final point (60_000, 8.2)
# First calculate the slope of the line by k = Δy / Δx
k = (8.2 - 14.1) / (60_000 - 4_000)

# Then we select the initial (or the final) point by y - y0 = k(x - x0). So y = kx + y0 - k*x0, which means, b = y0 - k*x0
b = 14.1 - k * 4_000

# Let's draw the graph and all of the points
fig, ax = plt.subplots(figsize=(6, 4.5))
ax.scatter(x=income, y=rate, label='Original Data')
ax.plot(income, k*income+b, 'b--', label='Fitted Line')
ax.set_xlabel('Income ($)')
ax.set_ylabel('Ulter Rate (Per 100 Population)')
ax.grid(True)
ax.legend()
plt.show()
```

##### (c) Find and graph the least squares regression line.

```python
def LSR_fit(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    X = np.column_stack([x, np.ones_like(x)])
    coeffs, *_ = np.linalg.lstsq(X, y, rcond=None)
    k, b = coeffs
    return k ,b
```

```python
k1, b1 = LSR_fit(x=income, y=rate)

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.scatter(x=income, y=rate, label='Original Data')
ax.plot(income, k*income+b, 'b--', label='Fitted Line')
ax.plot(income, k1*income+b1, 'r--', label='LSR Fitted Line')
ax.set_xlabel('Income ($)')
ax.set_ylabel('Ulter Rate (Per 100 Population)')
ax.grid(True)
ax.legend()
plt.show()
```

##### (d) Use the linear model in part (c) to estimate the ulcer rate for an income of $25,000.

```python
print(k1, b1)
rate_25000 = k1*25_000 + b1
print(f'Model in part (c) estimated ulcer rate for an income of $25,000 is {rate_25000:.2f}')

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.scatter(x=income, y=rate, label='Original Data')
ax.plot(income, k*income+b, 'b--', label='Fitted Line')
ax.plot(income, k1*income+b1, 'r--', label='LSR Fitted Line')
ax.scatter(x=25_000, y=rate_25000, marker='x', c='green', label='Model-c Predicted Rate for $25,000')
ax.set_xlabel('Income ($)')
ax.set_ylabel('Ulter Rate (Per 100 Population)')
ax.grid(True)
ax.legend()
plt.show()
```

##### (e) According to the model, how likely is someone with an income of $80,000 to suffer from peptic ulcers?

```python
rate_80000 = k1*80_000 + b1
print(f'According to the model in part (c), about {rate_80000:.2f} per 100 population with an income of $80,000 will suffer from peptic ulcers.')

extended_income = np.append(income, 80_000)

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.scatter(x=income, y=rate, label='Original Data')
ax.plot(extended_income, k*extended_income+b, 'b--', label='Fitted Line')
ax.plot(extended_income, k1*extended_income+b1, 'r--', label='LSR Fitted Line')
ax.scatter(x=80_000, y=rate_80000, marker='x', c='green', label='Model-c Predicted Rate for $25,000')
ax.set_xlabel('Income ($)')
ax.set_ylabel('Ulter Rate (Per 100 Population)')
ax.grid(True)
ax.legend()
plt.show()
```

##### (a) Fit a power model to the data.

> Let's have a look at the original points first.

```python
d = np.array([0.387, 0.723, 1.000, 1.523, 5.203, 9.541, 19.190, 30.086])
T = np.array([0.241, 0.615, 1.000, 1.881, 11.861, 29.457, 84.008, 164.784])

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.scatter(d, T, label='Original Data')
ax.set_xlabel('Average Distance (d) of The Planets From The Sun')
ax.set_ylabel('Periods of The Planets (T)')
ax.grid(True)
ax.legend()
plt.show()
```

> Review that a function of the form $f(x) = x^a$, where $a$ is a constant, is called a **power function**.

> The only way we know to fit a model is using the least square regression to fit a linear model. So we have to try to transport power model into linear model.
> 
> Fortunately, we find that: 
> $$ \begin{aligned}
> y &= x^a \\
> \log(y) &= \log(x^a) \\
> \log(y) &= a\log(x)
> \end{aligned} $$
> If we write $u = \log(y)$ and $v = \log(x)$, then we have $u = av$, which is a linear model!

```python
new_x = np.log(d)
new_y = np.log(T)

k32, b32 = LSR_fit(x=new_x, y=new_y)
print(f'The slope: {k32:.4f}, and the y-intercept: {b32:.4f}')
print(f'We could find that the y-intercept is quite close to zero. \n The fact fits out model and indicates that our assumption is suitable.')
```

> Let's draw the figure of the transported points and the linear model.

```python
x = np.linspace(new_x[0], new_x[-1], 100)

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.scatter(new_x, new_y, label='Transported Data')
ax.plot(x, k32*x+b32, 'g--', label='Fitted Linear Model')
ax.set_xlabel('Logarithm of The Distance')
ax.set_ylabel('Logarithm of The Periods')
ax.grid(True)
ax.legend()
plt.show()
```

> The linear model fits well! Let's just abandon the y-intercept since we assume the model to be power model, which mean the corresponding linear model has zero y-intercept.
> 
> According to the derivation above, the fitted linear model is $u = 1.4995v$. So the fitted power model is $y = x^{1.4995}$. Let draw it out.

```python
x = np.linspace(0, 30, 1000)

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.scatter(d, T, label='Original Data')
ax.plot(x, x**1.4995, 'g--', label='Fitted Power Model')
ax.set_xlabel('Average Distance (d) of The Planets From The Sun')
ax.set_ylabel('Periods of The Planets (T)')
ax.grid(True)
ax.legend()
plt.show()
```

> The domain of $f(x)$ is $(-\infty, 0) \cup (0, \infty)$, and the domain of $g(x)$ is $(-\infty, -2) \cup (-2，\infty)$. 
> 
> Since $g(x)$ should be in the domain of $f(x)$, which means $g(x) \neq 0$, that is $x\neq 1$, the domain of $f\circ g$ is $x\neq -2$ and $x\neq -1$, which is $(-\infty, -2) \cup (-2, -1) \cup (-1, \infty)$.
> 
> $$ \begin{aligned}
> f \circ g &= f(g(x)) \\
> &= g(x) + \frac{1}{g(x)} \\ 
> &= \frac{x+1}{x+2} + \frac{x+2}{x+1} \\ 
> &= \frac{2x^2 + 6x + 5}{(x+2)(x+1)}
> \end{aligned} $$
> 
> So $f\circ g = \frac{2x^2 + 6x + 5}{(x+2)(x+1)}$ ($x \neq -2, -1$).

##### (b) $g \circ f$

> Since $f(x)$ should be in the domain of $g(x)$, which means $f(x) \neq -2$, that is $x \neq -1$, the domain of $g \circ f$ is $x \neq 0$ and $x \neq -1$, which is $(-\infty, -1) \cup (-1, 0) \cap (0, \infty)$.
> 
> $$ \begin{aligned}
> g \circ f &= g(f(x)) \\
> &= \frac{f(x)+1}{f(x)+2} \\ 
> &= \frac{x + \frac{1}{x} + 1}{x + \frac{1}{x} + 2} \\ 
> &= \frac{x^2 + x + 1}{x^2 + 2x + 1} \\
> &= \frac{x^2 + x + 1}{(x+1)^2}
> \end{aligned} $$
> 
> So $f\circ g = \frac{x^2 + x + 1}{(x+1)^2}$ ($x \neq -1, 0$).

##### (c) $f \circ f$

> Since $f(x)$ should be in the domain of $f(x)$, which means $f(x) \neq 0$, but all of the $x\neq 0$ satisfy this restriction. So thee domain of $f\circ f$ is $x\neq 0$, which is $(-\infty, 0) \cup (0, \infty)$.
> 
> $$ \begin{aligned}
> f \circ f &= f(f(x)) \\
> &= f(x) + \frac{1}{f(x)} \\
> &= x + \frac{1}{x} + \frac{1}{x + \frac{1}{x}} \\
> &= x + \frac{1}{x} + \frac{x}{x^2+1} \\
> &= \frac{x^4 + 2x^2 + x + 1}{x(x^2+1)}
> \end{aligned}$$
> 
> So $f \circ f = \frac{x^4 + 2x^2 + x + 1}{x(x^2+1)}$ ($x\neq 0$).

##### (d) $g \circ g$

> Since $g(x)$ should be in the domain of $g(x)$, which means $g(x) \neq -2$, that is $x \neq -\frac{5}{3}$, the domain of $g\circ g$ is $x \neq -2$ and $x \neq -\frac{5}{3}$, which is $(-\infty, -2) \cup (-2, -\frac{5}{3}) \cup (-\frac{5}{3}, \infty)$.
> 
> $$\begin{aligned}
> g \circ g &= g(g(x)) \\
> &= \frac{g(x) + 1}{g(x) + 2} \\
> &= \frac{\frac{x+1}{x+2} + 1}{\frac{x+1}{x+2}+2} \\
> &= \frac{2x + 3}{3x + 5} \\
> \end{aligned}$$
> 
> So $g\circ g = \frac{2x+3}{3x+5}$ ($x\neq -2, -\frac{5}{3}$).

#### 66. Suppose $g$ is an odd function and let $h = f\circ g$. Is $h$ always an odd function? What if $f$ is odd? What if $f$ is even?

> 
> $$\begin{aligned}
> h(-x) &= f(g(-x)) \\
> &= f(-g(x))
> \end{aligned}$$
> In this case, we have no idea whether $h$ is odd or not.

>

> \end{aligned}$$
> So $h$ is always odd.

>

> h(-x) &= f(-g(x)) \\ 
> &= f(g(x)) = h(x)
> \end{aligned}$$
> So $h$ is always even.

#### 40. By comparing Definitions 2, 3, and 4, prove Theorem 1.6.1.

> **Definition 2: Precise Definition of a Limit** Let $f$ be a function defined on some open interval that contains the number $a$, except possibly at $a$ itself. Then we say that the limit of $f(x)$ as $x$ approaches $a$ is $L$, and we write 
> $$ \lim_{x\to a}f(x) = L $$
> if for every number $\epsilon > 0$ there is a number $\delta > 0$ such that 
> $$ \text{if } 0 < |x - a| < \delta \text{ then } |f(x) - L| < \epsilon $$

> **Definition 3: Definition of Left-Hand Limit**
> $$ \lim_{x\to a^-} f(x) = L $$
> if for every number $\epsilon > 0$ there is a number $\delta > 0$ such that 
> $$ \text{if } a - \delta < x < a \text{ then } |f(x) - L| < \epsilon $$

> **Definition 4: Definition of Right-Hand Limit**
> $$ \lim_{x\to a^+} f(x) = L $$
> if for every number $\epsilon > 0$ there is a number $\delta > 0$ such that 
> $$ \text{if } a < x < a + \delta \text{ then } |f(x) - L| < \epsilon $$

> **Theorem 1.6.1: Theorem** $\lim\limits_{x\to a}f(x) = L$ if and only if $\lim\limits_{x\to a^-}f(x) = L = \lim\limits_{x\to a^+}f(x)$

> i) For the sufficiency
> 
> If the left- and right-hand limits of $f(x)$ both exist and are the same value, for every number $\epsilon > 0$ there is number $\delta > 0$ such that: 
> $$ \begin{cases}
> |f(x) - L| < \epsilon & \text{if } a - \delta < x < a \\
> |f(x) - L| < \epsilon & \text{if } a < x < a + \delta
> \end{cases} $$
> which means, $|f(x) - L| < \epsilon \text{ if } a - \delta < x < a + \delta \text{ and } x \neq a$.
> 
> So if the left- and right hand limits of $f(x)$ both exists and are the same value, for every number $\epsilon > 0$ there is number $\delta > 0$ such that $|f(x) - L| < \epsilon \text{ if } 0 < |x - a| < \delta $. According to Definition 2, $\lim\limits_{x\to a}f(x) = L$.
>
> So $\lim\limits_{x\to a} f(x) = L$ if $\lim\limits_{x\to a^-}f(x) = L = \lim\limits_{x\to a^+}f(x)$.

> ii) For the necessity
> 
> If the limit of $f(x)$ at $x = a$ exists, for every number $\epsilon > 0$ there is number $\delta > 0$ such that $|f(x) - L| < \epsilon$ if $0 < |x - a| < \delta$. Notice that:
> $$ \begin{aligned}
> 0 < |&x - a| < \delta \\
> 0 < |x - a| &\text{ and } |x - a| < \delta \\
> x \neq a &\text{ and } a - \delta < x < a + \delta \\
> a - \delta < x < a &\text{ and } a < x < a + \delta
> \end{aligned} $$
> 
> So the limit could be stated like for every number $\epsilon > 0$ there is number $\delta > 0$ such that $|f(x) - L| < \epsilon$ if $ a - \delta < x < a$ and $|f(x) - L| < \epsilon$ if $a < x < a + \delta$. And according to Definition 3 and 4, the statement indicates that the left- and right-hand limits of $f(x)$ both exist and are the same value.
> 
> So $\lim\limits_{x\to a^-}f(x) = L = \lim\limits_{x\to a^+}f(x)$ if $\lim\limits_{x\to a}f(x) = L$.

> In summary, $\lim\limits_{x\to a}f(x) = L$ if and only if $\lim\limits_{x\to a^-}f(x) = L = \lim\limits_{x\to a^+}f(x)$.
