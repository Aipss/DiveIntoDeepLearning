> *51-56: Find an expression for the function whose graph is the given curve.*
#### 53. The bottom half of the parabola $x + (y-1)^2 = 0$

> Rewrite the equation $x + (y-1)^2 = 0$ in terms of $y$:
> $$ \begin{align*}
> x + (y-1)^2 &= 0 \\
> (y-1)^2 &= -x \\
> y-1 &= \pm \sqrt{-x} \\
> y &= 1 \pm \sqrt{-x}
> \end{align*} $$
> The bottom half is $y = 1 - \sqrt{-x}$, whose domain is $(-\infty, 0]$ and range is $(-\infty, 1]$. The graph of the function is shown below.

```python
x = np.linspace(-10, 0, 100)
y = 1 - np.sqrt(-x)

fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(x, y)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True)
plt.show()
```

#### 67. In a certain Country, income tax is assessed as follows. There is no tax on income up to $10,000. Any income over $10,000 is taxed at a rate of 10%, up to an income of $20,000. Any income over $20,000 is taxed at 15%.
##### (a) Sketch the graph of the tax rate $R$ as a function of the income $I$.

> According to the problem, we can define the tax rate function $R(I)$ as follows:
> $$ R(I) = \begin{cases}
> 0, & \text{if } 0 \leq I \leq 10,000 \\
> 0.10, & \text{if } 10,000 < I \leq 20,000 \\
> 0.15, & \text{if } I > 20,000
> \end{cases} $$
>
> So we could sketch the graph as follows:

```python
x1 = np.linspace(0, 10_000)
x2 = np.linspace(10_000, 20_000)
x3 = np.linspace(20_000, 42_000)
# x = np.concatenate([x1, x2, x3])

y1 = np.zeros_like(x1)
y2 = 0.1 * np.ones_like(x2)
y3 = 0.15 * np.ones_like(x3)
# y = np.concatenate([y1, y2, y3])

fig, ax = plt.subplots(figsize=(8, 6))
# ax.plot(x, y)

ax.plot(x1, y1, color='b')
ax.scatter(x1[0], y1[0], 
           marker='o', 
           facecolors='b', edgecolors='b')
ax.scatter(x1[-1], y1[-1], 
           marker='o', 
           facecolors='b', edgecolors='b')

ax.plot(x2, y2, color='b')
ax.scatter(x2[0], y2[0], 
           marker='o', 
           facecolors='none', edgecolors='b')
ax.scatter(x2[-1], y2[-1], 
           marker='o', 
           facecolors='b', edgecolors='b')

ax.plot(x3, y3, color='b')
ax.scatter(x3[0], y3[0], 
           marker='o', 
           facecolors='none', edgecolors='b')

ax.set_title("The tax rate of the income")
ax.set_ylabel('Tax Rate (R)')
ax.set_xlabel('Income (I)')
ax.grid(True)
plt.show()
```

> 
> Cause $\$20,000 \lt \$26,000$, the tax is $0.15 \times (\$26,000 - \$20,000) + 0.10 \times (\$20,000 - \$10,000) = \$1,900$.

##### (c) Sketch the graph of the total assessed tax $T$ as a function of the income $I$.

> $$ T(I) = 
> \begin{cases}
> 0, & \text{if } 0 \le I \le 10,000 \\
> R(I) * (I - 10,000), & \text{if } 10,000 \lt I \le 20,000 \\
> R(I) * (I - 20,000) + 0.1 * (20,000 - 10,000), & \text{if } I \gt 20,000
> \end{cases} $$
> That is
> $$ T(I) = 
> \begin{cases}
> 0, & \text{if } 0 \le I \le 10,000 \\

> *73-78: Determine whether $f$ is even, odd, or neither. If you have a graphing calculator, use it to check your answer visually.*
#### 78. $ f(x) = 1 + 3x^3 - x^5 $

> $$ \begin{align*}
> f(-x) &= 1 + 3(-x)^3 - (-x)^5 \\
>  &= 1 - 3x^3 + x^5 
> \end{align*}$$
>
> Because $f(x) + f(-x) = 2 \neq 0$ and $f(x) - f(-x) = 6x^3 - 2x^5 \neq 0$, 
>
> we could know that $f(x) \neq f(-x) $ and $f(x) \neq -f(-x)$.
>
> So $f(x)$ is neither even or odd.

```python
x = np.linspace(-1, 1, 100)
y = 1 + 3 * x**3 - x**5

fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(x, y)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')

ax.grid(True)
ax.text(1.05, 0.1, 'x', fontsize=14, va='center')
ax.text(0.05, 3.1, 'y', fontsize=14, ha='center')

plt.show()
```

#### 79. If $f$ and $g$ are both even functions, if $f+g$ even? If $f$ and $g$ are both odd functions, if $f+g$ odd? What if $f$ is even and $g$ is odd? Justify your answers.

> - **If $f$ and $g$ are both even functions:** 
> 
>   We have $f(x) = f(-x)$ and $g(x) = g(-x)$.   
> 
>   Let $h = f + g$, then $h(-x) = f(-x) + g(-x) = f(x) + g(x) = h(x)$, which indicates that $h$, that is $f+g$, is even.

> - **If $f$ and $g$ are both odd functions**
> 
>   We have $f(x) = -f(-x)$ and $g(x) = -g(-x)$.  
>  
>   Let $h = f+g$, then $h(-x) = f(-x) + g(-x) = -f(x) - g(x) = -h(x)$, which indicates that $h$, that is $f+g$, is odd.

> - **If $f$ is even and $g$ is odd:**
> 
>   We have $f(x) = f(-x)$ and $g(x) = -g(-x)$.   
> 
>   Let $h = f+g$, then $h(-x) = f(-x) + g(-x) = f(x) - g(x)$.   
> 
>   Because $h(x) + h(-x) = 2f(x)$ and $h(x) - h(-x) = 2g(x)$, so $h$, that is $f+g$, is even AND odd if and only if $f(x) \equiv g(x) \equiv 0$.    
>
>   Notice that $h \equiv 0$ when $f \equiv g \equiv 0$.

#### 80. If $f$ and $g$ are both even functions, is the product $fg$ even? If $f$ and $g$ are both odd functions, is $fg$ odd? What if $f$ is even and $g$ is odd? Justify your answers.

> - **If $f$ and $g$ are both even functions:**
>
>   We have $f(x) = f(-x)$ and $g(x) = g(-x)$.
> 
>   Let $h = fg$, then $h(-x) = f(-x)g(-x) = f(x)g(x) = h(x)$.
>
>   So $h$, that is $fg$, is even.

> - **If $f$ and $g$ are both odd functions:**
> 
>   We have $f(x) = -f(-x)$ and $g(x) = -g(-x)$.
> 
>   Let $h = fg$, then $h(-x) = f(-x)g(-x) = [-f(x)][-g(x)]=f(x)g(x) = h(x)$.
> 
>   So $h$, that is $fg$, is even.

| Income | Ulcer rate (Per 100 population) |
| :--: | :--: |
|$4,000|14.1|
|$6,000|13.0|

|$16,000|12.0|

|$30,000|10.5|
|$45,000|9.4|

##### (a) Make a scatter plot of these data and decide whether a linear model is appropriate.

```python
income = np.array([4, 6, 8, 12, 16, 20, 30, 45, 60]) * 1_000
rate = np.array([14.1, 13.0, 13.4, 12.5, 12.0, 12.4, 10.5, 9.4, 8.2])

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(x=income, y=rate)
ax.set_xlabel('Income ($)')
ax.set_ylabel('Ulcer Rate (Per 100 Population)')
ax.grid(True)
plt.show()
```

> According to the scatter plot above, we could see a line approximately.
> So we could give linear model a try.

##### (b) Find and graph a linear model using the first and last data points.

```python
# the line will go through the initial point (40_000, 14.1) and the final point (60_000, 8.2)
# First calculate the slope of the line by k = Δy / Δx
k = (8.2 - 14.1) / (60_000 - 40_000)

# Then we select the initial (or the final) point by y - y0 = k(x - x0). So y = kx + y0 - k*x0, which means, b = y0 - k*x0
b = 14.1 - k * 40_000

# Let's draw the graph and all of the points
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(x=income, y=rate)
ax.plot(income, k*income+b)
ax.set_xlabel('Income ($)')
ax.set_ylabel('Ulter Rate (Per 100 Population)')
ax.grid(True)
plt.show()
```
