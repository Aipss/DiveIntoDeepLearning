#### (a) If $g(x) = x^{2 / 3}$, show that $g'(0)$ does not exist. 

> $$\begin{aligned}
> g'(0) &= \lim_{h\to 0} \frac{g(0 + h) - g(0)}{h} \\
> &= \lim_{h\to 0} \frac{\sqrt[3]{(0 + h)^2} - \sqrt[3]{0^2}}{h} \\
> &= \lim_{h\to 0} \frac{\sqrt[3]{h^2}}{h} = \lim_{h\to 0} \frac{1}{\sqrt[3]{h}} \\
> \end{aligned}$$
>
> Thus $g'(0)$ does not exist.

#### (b) If $a\neq 0$, find $g'(a)$.

> $$\begin{aligned}
> g'(a) &= \lim_{h\to 0} \frac{g(a + h) - g(a)}{h} \\
> &= \lim_{h\to 0} \frac{\sqrt[3]{(a + h)^2} - \sqrt[3]{a^2}}{h} \\
> &= \lim_{h\to 0} \frac{(\sqrt[3]{(a + h)^2} - \sqrt[3]{a^2})(\sqrt[3]{(a + h)^2} + \sqrt[3]{a^2})}{h(\sqrt[3]{(a + h)^2} + \sqrt[3]{a^2})} \\
> &= \lim_{h\to 0} \frac{(a + h)^2 - a^2}{h(\sqrt[3]{(a + h)^2} + \sqrt[3]{a^2})} \\
> &= \lim_{h\to 0} \frac{2ah + h^2}{h(\sqrt[3]{(a + h)^2} + \sqrt[3]{a^2})} \\
> &= \lim_{h\to 0} \frac{2a + h}{\sqrt[3]{(a + h)^2} + \sqrt[3]{a^2}} \\
> &= \lim_{h\to 0} \frac{2a}{2\sqrt[3]{a^2}} = \frac{a}{\sqrt[3]{a^2}} = \sqrt[3]{a}
> \end{aligned}$$
>
> Thus if $a\neq 0$, $g'(a) = \sqrt[3]{a}$.

#### (c) Show that $y = x^{2/3}$ has a vertical tangent line at $(0, 0)$.

> Based on part (a), 
> $$g'(0) = \lim\limits_{h\to 0}\frac{1}{\sqrt[3]{h}}$$
>
> Thus 
> $$\lim_{h\to 0^+} \frac{1}{\sqrt[3]{h}} = +\infty$$
> $$\lim_{h\to 0^-} \frac{1}{\sqrt[3]{h}} = -\infty$$
>
> Which means that the slope of the tangent line at $(0, 0)$ is vertical.

#### (d) Illustrate part (c) by graphing $y = x^{2 / 3}$.

```python
x = np.linspace(-1, 1, 100000)
y = (x**2)**(1/3)

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.plot(x, y, label=r'$g(x) = x^{2/3}$')

ax.set_xlabel(r'$x$')
ax.set_ylabel(r'$y$')
ax.set_xlim(-1, 1)
ax.set_ylim(0, 1)
ax.grid(True)
ax.legend()
plt.show()
```

### 61. Recall that a function $f$ is called *even* if $f(-x) = f(x)$ for all $x$ in its domain and *odd* if $f(-x) = -f(x)$ for all such $x$. Prove each of the following.

#### (a) The derivative of an even function is an odd function.

> If $f(x)$ is even, $f(-x) = f(x)$. Since $f'(x) = \lim\limits_{h\to 0}\frac{f(x + h) - f(x)}{h}$ if it exists, we have
> $$\begin{aligned}
> f'(-x) &= \lim_{h\to 0}\frac{f(-x+h) - f(-x)}{h} \\
> &= \lim_{h\to 0}\frac{f(x - h) - f(x)}{h} \\ 
> &= \lim_{-h\to 0}\frac{f(x + h) - f(x)}{-h} \\
> &= \lim_{h\to 0}-\frac{f(x+h) - f(x)}{h} = -f'(x)
> \end{aligned}$$
>
> Thus the derivative of an even function is an odd function if it exists.

#### (b) The derivative of an odd function is an even function.

> If $f(x)$ is odd, $f(-x) = -f(x)$. Since $f'(x) = \lim\limits_{h\to 0}\frac{f(x + h) - f(x)}{h}$ if it exists, we have
> $$\begin{aligned}
> f'(-x) &= \lim_{h\to 0}\frac{f(-x+h) - f(-x)}{h} \\
> &= \lim_{h\to 0}\frac{-f(x - h) + f(x)}{h} \\ 
> &= \lim_{-h\to 0}\frac{-f(x + h) + f(x)}{-h} \\
> &= \lim_{h\to 0}-\frac{f(x+h) - f(x)}{h} = f'(x)
> \end{aligned}$$
>
> Thus the derivative of an odd function is an even function if it exists.
