#### (a) If $g(x) = x^{2 / 3}$, show that $g'(0)$ does not exist. 

> $$\begin{aligned}
> g'(0) &= \lim_{h\to 0} \frac{g(0 + h) - g(0)}{h} \\
> &= \lim_{h\to 0} \frac{\sqrt[3]{(0 + h)^2} - \sqrt[3]{0^2}}{h} \\
> &= \lim_{h\to 0} \frac{\sqrt[3]{h^2}}{h} = \lim_{h\to 0} \frac{1}{\sqrt[3]{h}} \\
> \end{aligned}$$
>
> Thus $g'(0)$ does not exist.

#### (b) If $a\neq 0$, find $g'(a)$.

> $$\begin{aligned}
> g'(a) &= \lim_{h\to 0} \frac{g(a + h) - g(a)}{h} \\
> &= \lim_{h\to 0} \frac{\sqrt[3]{(a + h)^2} - \sqrt[3]{a^2}}{h} \\
> &= \lim_{h\to 0} \frac{(\sqrt[3]{(a + h)^2} - \sqrt[3]{a^2})(\sqrt[3]{(a + h)^2} + \sqrt[3]{a^2})}{h(\sqrt[3]{(a + h)^2} + \sqrt[3]{a^2})} \\
> &= \lim_{h\to 0} \frac{(a + h)^2 - a^2}{h(\sqrt[3]{(a + h)^2} + \sqrt[3]{a^2})} \\
> &= \lim_{h\to 0} \frac{2ah + h^2}{h(\sqrt[3]{(a + h)^2} + \sqrt[3]{a^2})} \\
> &= \lim_{h\to 0} \frac{2a + h}{\sqrt[3]{(a + h)^2} + \sqrt[3]{a^2}} \\
> &= \lim_{h\to 0} \frac{2a}{2\sqrt[3]{a^2}} = \frac{a}{\sqrt[3]{a^2}} = \sqrt[3]{a}
> \end{aligned}$$
>
> Thus if $a\neq 0$, $g'(a) = \sqrt[3]{a}$.

> Thus 
> $$\lim_{h\to 0^+} \frac{1}{\sqrt[3]{h}} = +\infty$$
> $$\lim_{h\to 0^-} \frac{1}{\sqrt[3]{h}} = -\infty$$

### 61. Recall that a function $f$ is called *even* if $f(-x) = f(x)$ for all $x$ in its domain and *odd* if $f(-x) = -f(x)$ for all such $x$. Prove each of the following.

#### (a) The derivative of an even function is an odd function.

> If $f(x)$ is even, $f(-x) = f(x)$. Since $f'(x) = \lim\limits_{h\to 0}\frac{f(x + h) - f(x)}{h}$ if it exists, we have
> $$\begin{aligned}
> f'(-x) &= \lim_{h\to 0}\frac{f(-x+h) - f(-x)}{h} \\
> &= \lim_{h\to 0}\frac{f(x - h) - f(x)}{h} \\ 
> &= \lim_{-h\to 0}\frac{f(x + h) - f(x)}{-h} \\
> &= \lim_{h\to 0}-\frac{f(x+h) - f(x)}{h} = -f'(x)
> \end{aligned}$$
>
> Thus the derivative of an even function is an odd function if it exists.

#### (b) The derivative of an odd function is an even function.

> If $f(x)$ is odd, $f(-x) = -f(x)$. Since $f'(x) = \lim\limits_{h\to 0}\frac{f(x + h) - f(x)}{h}$ if it exists, we have
> $$\begin{aligned}
> f'(-x) &= \lim_{h\to 0}\frac{f(-x+h) - f(-x)}{h} \\
> &= \lim_{h\to 0}\frac{-f(x - h) + f(x)}{h} \\ 
> &= \lim_{-h\to 0}\frac{-f(x + h) + f(x)}{-h} \\
> &= \lim_{h\to 0}-\frac{f(x+h) - f(x)}{h} = f'(x)
> \end{aligned}$$
>
> Thus the derivative of an odd function is an even function if it exists.

#### (a) If $F(x) = f(x)g(x)$, where $f$ and $g$ have derivatives of all orders, show that $F'' = f''g + 2f'g' + fg''$.

> According to the Product Rule: 
> $$\begin{aligned}
> F' = \frac{dF}{dx} &= \frac{d}{dx}[f(x)g(x)] \\
> &= f(x)\frac{d}{dx}[g(x)] + g(x)\frac{d}{dx}[f(x)] \\
> &= fg' + gf'
> \end{aligned}$$
>
> So according to the Sum Rule and the Product Rule:
> $$\begin{aligned}
> F'' &= \frac{d(F')}{dx} = \frac{d^2F}{dx^2} \\ 
> &= \frac{d}{dx}[fg' + gf'] = \frac{d}{dx}(f(x)\frac{d}{dx}[g(x)] + g(x)\frac{d}{dx}[f(x)]) \\
> &= \frac{d}{dx}[fg'] + \frac{d}{dx}[gf'] \\
> &= (g'\frac{d}{dx}f + f\frac{d}{dx}g') + (f'\frac{d}{dx}g + g\frac{d}{dx}f') \\
> &= (g'f' + fg'') + (f'g' + gf'') \\
> &= f''g + 2f'g' + fg''
> \end{aligned}$$
>
> To sum up, $F'' = f''g + 2f'g' + fg''$.

> Since 
> $$\begin{aligned}
> \binom{n}{k} + \binom{n}{k+1} &= \frac{n!}{k!(n-k)!} + \frac{n!}{(k+1)!(n-k-1)!} \\
> &= \frac{n!}{k!(n-k)!} + \frac{n!\times (n-k)}{k!(k+1)(n-k)!} \\
> &= \frac{n!}{k!(n-k)!}\times (1 + \frac{n-k}{k+1}) \\
> &= \frac{n!}{k!(n-k)!}\frac{n+1}{k+1} \\
> &= \frac{(n+1)!}{(k+1)!(n-k)!} \\
> &= \frac{(n+1)!}{(k+1)![(n+1) - (k+1)]!} = \binom{n+1}{k+1}
> \end{aligned}$$
>
> Thus
> $$\begin{aligned}
> (2) &= \binom{n}{0}f^{(n+1)}g + \binom{n+1}{1}f^{(n)}g^{(1)} + \cdots + \binom{n+1}{k}f^{(n-k+1)}g^{(k)} + \binom{n+1}{k+1}f^{(n-k)}g^{(k+1)} + \\
> & \;\;\;\; \cdots + \binom{n+1}{n}f^{(1)}g^{(n)} + \binom{n}{n}f^{(n+1)}g \\
> &= \binom{n+1}{0}f^{(n+1)}g + \binom{n+1}{1}f^{(n+1-1)}g^{(1)} + \cdots + \binom{n+1}{k}F^{(n+1-k)}g^{(k)} + \cdots + \binom{n+1}{n}f^{(1)}g^{(n+1-1)} + \binom{n+1}{n+1}fg^{(n+1)}
> \end{aligned}$$
> which satisfies the formula $(1)$.

> So formula $(1)$ is what we want for $F^{(n)}$.

> ---

### 105. An easy proof of the Quotient Rule can be given if we make the prior assumption that $F'(x)$ exists, where $F = f/g$. Write $f = Fg$; then differentiate using the Product Rule and solve the resulting equation for $F'$.

> According to the Product Rule, 
> $$f' = \frac{d}{dx}[Fg] = g\frac{d}{dx}F + F\frac{d}{dx}g = F'g + Fg'$$
>
> Since $F = f/g$, $g$ is non-zero in the domain of $F$. So
> $$F' = \frac{f' - Fg'}{g} = \frac{f' - \frac{f}{g}g'}{g} = \frac{f'g - fg'}{g^2}$$
> which satisfies the Quotient Rule.

## 2.4 Derivatives of Trigonometric Functions

> ---

### 17. Prove that $\frac{d}{dx}(\csc x) = -\csc x \cot x$.

> $$\begin{aligned}
> \frac{d}{dx}(\csc x) = \frac{d}{dx}\frac{1}{\sin x} &= (-1)\frac{1}{\sin^2 x}\times \cos x \\
> &= -\frac{1}{\sin x}\times \frac{\cos x}{\sin x} = -\csc x \cot x
> \end{aligned}$$

> ---

### 19. Prove that $\frac{d}{dx}(\cot x) = -\csc^2 x$.

> $$\begin{aligned}
> \frac{d}{dx}(\cot x) = \frac{d}{dx}\frac{\cos x}{\sin x} &= \frac{\sin x\frac{d}{dx}\cos x - \cos x\frac{d}{dx}\sin x}{\sin^2 x} \\ 
> &= \frac{\sin x (-\sin x) - \cos x \cos x}{\sin^2 x} \\
> &= -\frac{\sin^2 x + \cos^2 x}{\sin^2 x} \\
> &= -\frac{1}{\sin^2 x} = -\csc^2 x
> \end{aligned}$$

> ---

### 20. Prove, using the definition of derivative, that if $f(x) = \cos x$, then $f'(x) = -\sin x$.

> $$\begin{aligned}
> f'(x) = \lim_{h\to 0}\frac{f(x+h) - f(x)}{h} &= \lim_{h\to 0}\frac{\cos (x+h) - \cos x}{h} \\
> &= \lim_{h\to 0}\frac{\cos(\frac{(x+h) + x}{2} + \frac{(x+h) - x}{2}) - \cos(\frac{(x+h) + x}{2} - \frac{(x+h) - x}{2})}{h} \\
> &= \lim_{h\to 0}\frac{-2\sin(x+\frac{h}{2})\sin\frac{h}{2}}{h} \\
> &= -\lim_{h\to 0}\sin(x + \frac{h}{2})\times\lim_{h\to 0}\frac{\sin\frac{h}{2}}{\frac{h}{2}} \\
> &= -\lim_{\frac{h}{2}\to 0}\sin(x + \frac{h}{2})\times\lim_{\frac{h}{2}\to 0}\frac{\sin\frac{h}{2}}{\frac{h}{2}} \\
> &= -\sin x \times 1 = -\sin x
> \end{aligned}$$
